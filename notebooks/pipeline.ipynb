```python
import sys
from pathlib import Path

PROJECT_ROOT = Path().resolve().parent
sys.path.append(str(PROJECT_ROOT))

from src.pipeline.core.survival.data import Data
from src.pipeline.config.survival_config import SurvivalConfig
from src.pipeline.utilities.data_loader import DataLoader
from src.pipeline.utilities.data_preparation import DataPreparation
from src.pipeline.config.imputer_config import ImputerConfig
from src.pipeline.core.experiment import ExperimentRunner
from src.pipeline.config.config_loader import load_survival_config, load_imputer_config
from src.pipeline.utilities.experiment_logger import ExperimentLogger

survival_config = SurvivalConfig()
data_loader = DataLoader()
data_preparation = DataPreparation()
imputer_config = ImputerConfig()

data = Data(
    config=survival_config,
    data_loader=data_loader,
    data_preparation=data_preparation,
)

config_path = PROJECT_ROOT / "yaml" / "xgb_cox.yaml"
imputer_config = load_imputer_config(config_path)
survival_config = load_survival_config(config_path)

# Initialize Logger
logger = ExperimentLogger(config=survival_config.base)
logger.save_config({
    "imputer": imputer_config.model_dump(),
    "survival": survival_config.model_dump()
})

experiment = ExperimentRunner(
    data=data,
    imputer_data_config=imputer_config.data,
    imputer_trainer_config=imputer_config.train,
    imputer_model_config=imputer_config.model,
    survival_config=survival_config,
    logger=logger,
)

# original_df_dict = data.load_original_data_dict()
prepared_df_dict = data.load_prepared_data_dict()

results = experiment.run_multiple(10)
```
